## 2. Fix the day/month swap

In the source, dates with day ≤ 12 were stored as `YYYY-DD-MM` (typical Excel locale swap).
For example `2014-01-12, 2014-02-12, 2014-03-12` are really 1, 2, 3 December 2014.

Evidence: compare the weekend/weekday ratio of the ambiguous rows (day ≤ 12) with the unambiguous rows (day > 12).

```python
iso = pd.to_datetime(df["ticket_date"], format="%Y-%m-%d")
ydm = pd.to_datetime(df["ticket_date"], format="%Y-%d-%m", errors="coerce")
ambiguous = iso.dt.day <= 12
print(f"Ambiguous rows: {ambiguous.sum()} of {len(df)} ({ambiguous.mean()*100:.1f} %)")
```

```python
def weekend_ratio(dates, mask):
    d = df.loc[mask, "net_weight_kg"].groupby(dates[mask]).sum()
    p = d.groupby(d.index.dayofweek).mean()
    return p[[5, 6]].mean() / p[[0, 1, 2, 3, 4]].mean()

rows = []
for y in range(2014, 2019):
    m = df["ticket_date"].str[:4] == str(y)
    rows.append({"year": y,
                 "unambiguous (day>12)": weekend_ratio(iso, m & ~ambiguous),
                 "ambiguous as stored (Y-M-D)": weekend_ratio(iso, m & ambiguous),
                 "ambiguous swapped (Y-D-M)": weekend_ratio(ydm, m & ambiguous)})
pd.DataFrame(rows).set_index("year").round(2)
```

Only the swapped reading reproduces the weekly pattern of the unambiguous rows. Apply the fix:

```python
df["ticket_date_original"] = df["ticket_date"]
```

```python
df["ticket_date"] = iso.where(~ambiguous, ydm)
```

```python
df["date_swapped"] = ambiguous & (iso.dt.month != iso.dt.day)   # e.g. 2015-03-03 is unchanged
```

## 4. Preprocessing

### 4.1 Calendar days
`ticket_date` is the collection date, and weekend values are real collections, so they are **kept as recorded**.
The calendar runs 2015-01-01 to 2018-12-31, as in the existing preprocessed file.
Every calendar day without a record becomes a gap to impute.

```python
START = pd.Timestamp("2015-01-01")
END = pd.Timestamp("2018-12-31")
```

```python
prep = daily.reindex(pd.date_range(START, END, freq="D")).to_frame("net_weight_kg")
prep.index.name = "ticket_date"
prep["net_weight_kg_raw"] = prep["net_weight_kg"]
prep["is_weekend"] = prep.index.dayofweek >= 5
prep["is_missing"] = prep["net_weight_kg_raw"].isna()
```

```python
print("Calendar:", START.date(), "to", END.date(), "|", len(prep), "days")
print("Days with recorded values:", int((~prep["is_missing"]).sum()), "| missing days:", int(prep["is_missing"].sum()))
print("Last recorded day:", daily.index.max().date(),
      "| days after it up to END:", int((prep.index > daily.index.max()).sum()))
```

### 4.2 Weekly pattern: how weekends are handled without setting them to 0
Weekends are about 20–30 % heavier than weekdays. Instead of removing that, the model learns it:
- `dayofweek` and `is_weekend` features tell the model which day it is predicting
- `lag_7`, `lag_14`, … give it the same weekday in previous weeks
- outliers are judged against the **same weekday** in nearby weeks, so a normal heavy Saturday is not flagged

```python
order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
obs = prep.loc[~prep["is_missing"], "net_weight_kg"]
obs.groupby(obs.index.day_name()).agg(["mean", "median", "count"]).reindex(order).round(0)
```

### 4.3 Outlier detection (same-weekday comparison)
Each recorded day is compared with the median of the same weekday over ±3 weeks (7 same-weekday values), using real values only.
Days with |robust z| > `Z_THRESHOLD` are flagged and imputed in 4.4. The raw value stays in `net_weight_kg_raw`.

```python
Z_THRESHOLD = 3.5

local_med = pd.Series(np.nan, index=prep.index)
local_mad = pd.Series(np.nan, index=prep.index)
for dow in range(7):
    s = prep.loc[prep.index.dayofweek == dow, "net_weight_kg"]
    med = s.rolling(7, center=True, min_periods=3).median()
    local_med.loc[s.index] = med
    local_mad.loc[s.index] = (s - med).abs().rolling(7, center=True, min_periods=3).median()

prep["robust_z"] = 0.6745 * (prep["net_weight_kg"] - local_med) / local_mad.replace(0, np.nan)
prep["is_outlier"] = prep["robust_z"].abs() > Z_THRESHOLD
print("Outliers flagged:", int(prep["is_outlier"].sum()),
      "| on weekends:", int((prep["is_outlier"] & prep["is_weekend"]).sum()))
prep.loc[prep["is_outlier"], ["net_weight_kg_raw", "robust_z"]].assign(
    weekday=lambda x: x.index.day_name(), same_weekday_median=local_med[prep["is_outlier"]]).round(1)
```

```python
fig, ax = plt.subplots()
ax.plot(prep.index, prep["net_weight_kg_raw"], lw=0.7, label="recorded kg")
o = prep[prep["is_outlier"]]
ax.scatter(o.index, o["net_weight_kg_raw"], color="red", zorder=3, label="outlier")
ax.legend(); ax.set_ylabel("kg"); ax.set_title("Outliers (compared with the same weekday)")
plt.show()
```

```python
# Outliers become gaps and are imputed together with the missing days
prep.loc[prep["is_outlier"], "net_weight_kg"] = np.nan
to_impute = prep["net_weight_kg"].isna()
print("Days to impute:", int(to_impute.sum()), f"({to_impute.mean() * 100:.1f} % of calendar days)",
      "| missing:", int(prep["is_missing"].sum()), "| outliers:", int(prep["is_outlier"].sum()))
```

### 4.4 Imputation with XGBoost

- trained **only on days with real values** (all weekdays, including weekends)
- features: lags 1..L, `dayofweek`, `is_weekend`, `month`. Lags that fall on gaps stay NaN (XGBoost handles them), so lag features never contain imputed values during training or testing
- time-ordered split of the real days: train 60 % / validation 20 % (choose settings) / test 20 % (final check)
- **validation and test contain only real recorded days, never imputed ones**
- compared with a seasonal-naive baseline (same weekday last week)

```python
def make_features(series, lag_count):
    X = pd.DataFrame(index=series.index)
    X = pd.concat([X] + [series.shift(i).rename(f"lag_{i}") for i in range(1, lag_count + 1)], axis=1)
    X["dayofweek"] = series.index.dayofweek
    X["is_weekend"] = (series.index.dayofweek >= 5).astype(int)
    X["month"] = series.index.month
    return X

def splits(index):
    n = len(index)
    return index[: int(n * 0.6)], index[int(n * 0.6): int(n * 0.8)], index[int(n * 0.8):]

target = prep["net_weight_kg"]                 # real values only; gaps and outliers are NaN
known_days = target.index[target.notna()]
train_idx, val_idx, test_idx = splits(known_days)
print("train / val / test real days:", len(train_idx), len(val_idx), len(test_idx))
print("test period:", test_idx.min().date(), "to", test_idx.max().date())
```

```python
lag_list = [7, 14, 28]
depth_list = [3, 5]
estimator_list = [200, 500]
eta_list = [0.03, 0.1]

results = []
for lag in lag_list:
    X = make_features(target, lag)
    for depth in depth_list:
        for n_est in estimator_list:
            for eta in eta_list:
                model = xgboost.XGBRegressor(n_estimators=n_est, max_depth=depth, learning_rate=eta,
                                             subsample=0.8, objective="reg:squarederror",
                                             random_state=0, n_jobs=-1)
                model.fit(X.loc[train_idx], target.loc[train_idx])
                pred = model.predict(X.loc[val_idx])
                results.append({"lag_count": lag, "max_depth": depth, "n_estimators": n_est, "eta": eta,
                                "val_rmse": sqrt(mean_squared_error(target.loc[val_idx], pred)),
                                "val_mae": mean_absolute_error(target.loc[val_idx], pred)})

results_df = pd.DataFrame(results).sort_values("val_rmse").reset_index(drop=True)
```

```python
results_df.head(10)
```

```python
best = results_df.iloc[0]
LAG = int(best["lag_count"])
print("Best settings:", best.to_dict())

X = make_features(target, LAG)
fit_idx = train_idx.append(val_idx)
best_model = xgboost.XGBRegressor(n_estimators=int(best["n_estimators"]), max_depth=int(best["max_depth"]),
                                  learning_rate=float(best["eta"]), subsample=0.8,
                                  objective="reg:squarederror", random_state=0, n_jobs=-1)
best_model.fit(X.loc[fit_idx], target.loc[fit_idx])
test_pred = best_model.predict(X.loc[test_idx])

y_test = target.loc[test_idx]
baseline = target.shift(7).loc[test_idx]
ok = baseline.notna()
print(f"Test  XGBoost   RMSE {sqrt(mean_squared_error(y_test, test_pred)):,.0f} | MAE {mean_absolute_error(y_test, test_pred):,.0f}")
print(f"Test  lag-7     RMSE {sqrt(mean_squared_error(y_test[ok], baseline[ok])):,.0f} | MAE {mean_absolute_error(y_test[ok], baseline[ok]):,.0f}")

plt.plot(y_test.index, y_test.values, label="actual")
plt.plot(y_test.index, test_pred, label="XGBoost")
plt.title("Test period (real recorded days only)"); plt.legend(); plt.show()
```

### 4.4.1 Is the reduced grid (24 configurations) good enough?

The original notebook searched 1,512 configurations. This notebook uses 24, because:
1. **Many original configurations were invalid or wasteful.** Lag 360 needs almost a year of history per prediction and broke the test set (RMSE = 0).
   10 trees with eta 0.01 cannot learn. Depths 6–9 overfit about 1,300 days of data.
2. **Searching more configurations overfits the validation set.** Every configuration is scored on the same validation days.
   With 1,512 tries, the winner is partly the luckiest one, so its validation score is too optimistic (Cawley & Talbot, 2010).
3. **Only a few hyperparameters matter much**, so a small, sensible grid usually finds close to the same optimum (Bergstra & Bengio, 2012).

This is checked here, not assumed. Both grids use the same train/validation split, and each winner is scored on the same untouched test set (real days only).
A bootstrap over the test days gives a 95 % confidence interval for the difference in test RMSE.

How to read it:
- **CI includes 0** (or the difference is small): the reduced grid is as good. Keep it.
- **CI entirely above 0** (the full grid is clearly better on test): widen the reduced grid.
- **Full grid better on validation but not on test**: that is validation overfitting (point 2) shown directly.

The full grid trains 1,512 models, so this cell is slow.

```python
# Refit on all known weekdays, then fill gaps in time order so later gaps can use earlier filled values
final_model = xgboost.XGBRegressor(n_estimators=int(best["n_estimators"]), max_depth=int(best["max_depth"]),
                                   learning_rate=float(best["eta"]), subsample=0.8,
                                   objective="reg:squarederror", random_state=0, n_jobs=-1)
final_model.fit(make_features(target, LAG).loc[known_days], target.loc[known_days])

filled = target.copy()
for day in filled.index[filled.isna()]:
    pos = filled.index.get_loc(day)
    row = {f"lag_{i}": (filled.iloc[pos - i] if pos - i >= 0 else np.nan) for i in range(1, LAG + 1)}
    row["dayofweek"] = day.dayofweek
    row["is_weekend"] = int(day.dayofweek >= 5)
    row["month"] = day.month
    filled.loc[day] = max(float(final_model.predict(pd.DataFrame([row]))[0]), 0.0)

prep["is_imputed"] = prep["net_weight_kg"].isna()
prep["net_weight_kg"] = filled
```

```python
imp = prep.loc[prep["is_imputed"], "net_weight_kg"]
print("Imputed values:", len(imp), f"({len(imp) / len(prep) * 100:.2f} % of all days)")
print(f"Mean imputed: {imp.mean():,.0f} kg | mean recorded: {prep.loc[~prep['is_imputed'], 'net_weight_kg'].mean():,.0f} kg")
print("Imputed per weekday:", prep.loc[prep["is_imputed"]].index.day_name().value_counts().reindex(order).to_dict())
print("Blanks left:", int(prep["net_weight_kg"].isna().sum()))
```

```python
fig, ax = plt.subplots()
ax.plot(prep.index, prep["net_weight_kg"], lw=0.8, label="final series")
i = prep[prep["is_imputed"]]
ax.scatter(i.index, i["net_weight_kg"], s=8, color="red", zorder=3, label="imputed")
ax.set_title("Final preprocessed daily series"); ax.set_ylabel("kg"); ax.legend()
plt.show()
```
